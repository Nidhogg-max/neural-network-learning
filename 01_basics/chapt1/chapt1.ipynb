## 1.1标量的创建及基本运算

```python
import torch
x1 = torch.tensor(3.0)
x2 = torch.tensor(2.0)
print(x1 , x2)
```

标量由只有一个元素的张量表示,在这里演示实例化两个标量并进行一些计算处理

```python
print(x1 + x2 , x1 * x2 , x1 - x2 , x1 / x2)
```

在这里我们如果定义两个标量是浮点型,那么同python一样,计算结果也均是浮点型

```python
#对x1,x2的定义进行修改试试
x1 = torch.tensor(3)
x2 = torch.tensor(2)
print(x1 , x2)
print(x1 + x2 , x1 * x2 , x1 - x2 , x1 / x2)
x2 = torch.tensor(2.0)
print(x2 , x1 + x2 , x1 * x2 , x1 - x2 , x1 / x2)
```

如果将x1,x2分别定义成整数3和2，那么无论是输出x1,x2还是进行计算后的结果，都为整数，**但值得注意的是，若执行的是除法运算，那么无论输入标量是否为整数，都将输出浮点数**，这样的做法主要是为了保证计算结果精度,并且若为整数与浮点数之间的计算，那么结果将输出浮点数

## 1.5降维

### 1.5.1求和降维法

```python
A5 = torch.arange(4 , dtype = torch.float32)
print(A5)
print(A5.sum())
A6 = torch.arange(24 , dtype = torch.float32).reshape(2 , 3 , 4)
print(A6)
print(A6.sum())
```

使用.sum()可以对张量进行求和，无论是几阶张量，都将降维成一个求和标量，但是使用求和符号sum只能使张量沿所有轴进行降维

### 1.5.2沿任意轴进行降维

```python
print(A4)
print(A4.shape)
A4_sum_axis0 = A4.sum(axis = 0)
print("沿行向量(轴0)降维:\n" , A4_sum_axis0)
print(A4_sum_axis0.shape)
A4_sum_axis1 = A4.sum(axis = 1)
print("沿列向量(轴1)降维:\n" , A4_sum_axis1)
print(A4_sum_axis1.shape)
A4_sum_axis0_axis1 = A1.sum(axis = [0 , 1])
print("沿着行向量和列向量(0轴和1轴)降维：\n" , A4_sum_axis0_axis1)
```

在通过求和符号进行实际降维的过程中可以通过axis进行指定沿何轴进行求和，对于二阶张量即矩阵，0轴对应行，1轴对应列,使用axis指定所有轴相当于直接使用sum求和符号进行降维

## 1.6非降维求和

对于降维求和，会减少张量的轴数，但是有时保留轴数也是必要的

```python
import torch
A4 = torch.arange(20 , dtype = torch.float32).reshape(5 , 4) 
print(A4)
sum_A4 = A4.sum(axis = 1 ,keepdims = True)  #keepdims = True即保留该轴的指令
print(sum_A4)
print(A4 / sum_A4)
print(A4.cumsum(axis = 0))
```

要想保留轴即使用keepdims = True指令，默认情况下keepdims = False，即降维求和。
对于非降维求和的重要作用主要体现在：想用每行的总和去除以该行的每个元素（即归一化），如果不保持维度，形状会不匹配；而保持维度后，A / sum_A 会利用广播机制自动对齐，计算简洁且正确。
或者使用cumsum函数，这个函数可以保证沿着张量的某个轴计算该张量的累积总和但不会沿着任何轴降低张量的维度，比如在cumsum内使用axis = 0，即按行计算

## 1.7点积

```python
A7 = torch.ones(4 , dtype = torch.float32)
A8 = torch.arange(4.0)
print(A7 , A8)
print(torch.dot(A7 , A8))
print(torch.sum(A7 * A8))
```

点积是两个矩阵相同位置元素的按元素乘积的和,并且，我们可以通过执行按照元素乘法，然后进行求和来表示向量的点积

## 1.8向量积与矩阵乘法

```python
A8 = torch.arange(4.0 , dtype = torch.float32)
A7 = A7.reshape(1 , 4)
print(A7.shape , A8.shape)
print(torch.mv(A7 , A8))
```

在代码中使用张量表示矩阵‐向量积，我们使用mv函数。当我们为矩阵A和向量x调用torch.mv(A7, A8)时，会执
行矩阵‐向量积。注意，A7的列维数（沿轴1的长度）必须与A8的维数（其长度）相同。

```python
A9 = torch.arange(12).reshape(3 , 4)
A10 = torch.arange(16).reshape(4 , 4)
print(torch.mm(A9 , A10))
```

在代码中实现了矩阵乘法，需要注意的是在进行矩阵乘法操作时，需要保证前一个矩阵的列数等于后一个矩阵的行数，即A9_axis1 == A10_axis0
